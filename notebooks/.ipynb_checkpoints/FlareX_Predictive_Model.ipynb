```python

import pandas as pd
import numpy as np
from pathlib import Path

# Load the existing FlareX FIRMS dataset
csv_path = Path(
    "/Users/manyamahajan/Desktop/projects/FlareX/data/raw/"
    "DL_FIRE_J2V-C2_837818/fire_nrt_J2V-C2_837818.csv"
)

df = pd.read_csv(csv_path)

# Standardise column names
df.columns = df.columns.str.strip().str.lower()

print("DATASET OVERVIEW")
print("----------------")
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nColumn names:")
print(df.columns.tolist())

print("\nFIRST FIVE RECORDS")
display(df.head())

# Check date coverage
if "acq_date" in df.columns:
    dates = pd.to_datetime(df["acq_date"], errors="coerce")
    print("\nDATE COVERAGE")
    print("Earliest:", dates.min())
    print("Latest:", dates.max())
    print("Unique dates:", dates.nunique())

# Geographic coverage
print("\nGEOGRAPHIC COVERAGE")
for col in ["latitude", "longitude", "frp"]:
    if col in df.columns:
        print(f"\n{col}:")
        print(pd.to_numeric(df[col], errors="coerce").describe())

# Sensor distribution
if "satellite" in df.columns:
    print("\nSATELLITE DISTRIBUTION")
    print(df["satellite"].value_counts())

# Missing values
print("\nMISSING VALUES")
print(df.isnull().sum())
```
